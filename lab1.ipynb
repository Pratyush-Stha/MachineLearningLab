```python
VOWEL_SET = set("aeiouAEIOU")


def classify_letters(text):
    letters_only = [ch for ch in text if ch.isalpha()]
    total_vowels = sum(1 for ch in letters_only if ch in VOWEL_SET)
    total_consonants = len(letters_only) - total_vowels
    return total_vowels, total_consonants


def run():
    text = input("Enter a string: ")
    n_vowels, n_consonants = classify_letters(text)

    print("Number of vowels:", n_vowels)
    print("Number of consonants:", n_consonants)


if __name__ == "__main__":
    run()
```

```python
def matmul(mat_a, mat_b):
    mat_b_cols = list(zip(*mat_b))

    product = [
        [sum(a_val * b_val for a_val, b_val in zip(row, col)) for col in mat_b_cols]
        for row in mat_a
    ]

    return product


def display_matrix(label, matrix):
    print(label)
    for row in matrix:
        print(list(row))


def run():

    mat_a = [
        [1, 2, 3],
        [4, 5, 6],
    ]


    mat_b = [
        [7, 8],
        [9, 10],
        [11, 12],
    ]


    if len(mat_a[0]) != len(mat_b):
        print("Error: Matrix multiplication is not possible.")
    else:
        product = matmul(mat_a, mat_b)

        display_matrix("Matrix A:", mat_a)
        print()
        display_matrix("Matrix B:", mat_b)
        print()
        display_matrix("Product Matrix (A × B):", product)


if __name__ == "__main__":
    run()
```

```python
def shared_values(seq_a, seq_b):
    lookup = set(seq_b)
    seen = set()
    overlap = []

    for value in seq_a:
        if value in lookup and value not in seen:
            overlap.append(value)
            seen.add(value)

    return overlap


def run():
    seq_a = list(map(int, input("Enter first list elements: ").split()))
    seq_b = list(map(int, input("Enter second list elements: ").split()))

    overlap = shared_values(seq_a, seq_b)

    print("Common elements:", overlap)
    print("Number of common elements:", len(overlap))


if __name__ == "__main__":
    run()
```

```python
def read_grid(n_rows):
    grid = []

    for _ in range(n_rows):
        row = list(map(int, input().split()))
        grid.append(row)

    return grid


def flip_matrix(grid):
    return [list(col) for col in zip(*grid)]


def run():
    n_rows = int(input("Enter number of rows: "))
    n_cols = int(input("Enter number of columns: "))

    print("Enter matrix:")
    grid = read_grid(n_rows)

    flipped = flip_matrix(grid)

    print("Transpose Matrix:")
    for row in flipped:
        print(row)


if __name__ == "__main__":
    run()
```

```python
import random
from collections import Counter



def sample_numbers(count, low, high):
    return [random.randint(low, high) for _ in range(count)]



def summarize_numbers(numbers):
    n = len(numbers)
    ordered = sorted(numbers)

    avg = sum(numbers) / n

    mid = n // 2
    if n % 2 == 0:
        med = (ordered[mid - 1] + ordered[mid]) / 2
    else:
        med = ordered[mid]

    tally = Counter(numbers)
    top_freq = max(tally.values())
    modes = [value for value, freq in tally.items() if freq == top_freq]

    return avg, med, modes


def run():
    numbers = sample_numbers(100, 100, 150)

    avg, med, modes = summarize_numbers(numbers)

    print("Generated Numbers:")
    print(numbers)

    print("\nMean:", avg)
    print("Median:", med)
    print("Mode:", modes)


if __name__ == "__main__":
    run()
```
