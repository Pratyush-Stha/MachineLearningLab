```python
def compute_average(values):
    '''Arithmetic mean, computed via the built-in sum()/len() instead
    of a manual accumulator loop.'''
    return sum(values) / len(values)


def compute_dispersion(values):
    '''Population variance using the E[x^2] - (E[x])^2 identity
    instead of accumulating squared deviations one at a time.'''
    n = len(values)
    if n == 0:
        return 0

    avg = compute_average(values)
    avg_of_squares = sum(v ** 2 for v in values) / n

    return avg_of_squares - avg ** 2
```

```python

import pandas as pd

raw_data = pd.read_csv("Lab Session Data(Purchase data).csv")
purchase_df = pd.DataFrame(raw_data)
purchase_df
```

```python
import numpy as np
import numpy.linalg as lin

purchase_matrix = np.loadtxt(
    "Lab Session Data(Purchase data).csv",
    delimiter=",",
    skiprows=1,
    usecols=(1, 2, 3, 4),
)

purchase_matrix
```

```python

col_headers = purchase_df.columns.to_numpy()

product_names = col_headers[1:]
print(product_names)
```

```python

qty_matrix = purchase_matrix[:, :3]


payment_vector = purchase_matrix[:, 3]


unit_costs, *_ = lin.lstsq(qty_matrix, payment_vector, rcond=None)

unit_costs
```

```python
for product, cost in zip(product_names, unit_costs):
    print(f"{product}: Rs. {cost:.2f}")
```

```python

stock_df = pd.read_csv("IRCTC Stock Price.csv")
stock_df
```

```python

clean_prices = pd.to_numeric(
    stock_df["Price"].str.replace(",", ""), errors="coerce"
).dropna().to_numpy()

clean_prices
```

```python
int(np.mean(clean_prices))
```

```python
int(np.var(clean_prices))
```

```python

print(compute_average(clean_prices))
```

```python

print(compute_dispersion(clean_prices))
```

```python

wed_subset = stock_df[stock_df["Day"] == "Wed"]


wed_prices = pd.to_numeric(
    wed_subset["Price"].str.replace(",", ""), errors="coerce"
).dropna().to_numpy()


avg_wed_price = compute_average(wed_prices)

print(f"Mean price on Wednesdays: {avg_wed_price:.2f}")
```

```python

apr_subset = stock_df[stock_df["Month"] == "Apr"]

apr_prices = pd.to_numeric(
    apr_subset["Price"].str.replace(",", ""), errors="coerce"
).dropna().to_numpy()

avg_apr_price = compute_average(apr_prices)

print(f"Mean price in April: {avg_apr_price:.2f}")
```

```python

chg_clean = pd.to_numeric(
    stock_df["Chg%"].str.rstrip("%"), errors="coerce"
).dropna()


is_loss = chg_clean < 0
n_losses = is_loss.sum()

loss_probability = is_loss.mean()

print(f"Total valid 'Chg%' entries: {len(chg_clean)}")
print(f"Number of negative 'Chg%' entries (losses): {n_losses}")
print(f"Probability of making a loss: {loss_probability:.4f}")
```

```python

wed_chg_pct = pd.to_numeric(
    stock_df[stock_df["Day"] == "Wed"]["Chg%"].str.rstrip("%"), errors="coerce"
).dropna()

total_wed_days = wed_chg_pct.size
wed_loss_probability = (wed_chg_pct < 0).mean()
wed_profit_probability = (wed_chg_pct > 0).mean()

print(f"Wednesday sample size: {total_wed_days}")
print(f"Probability of a loss on Wednesday: {wed_loss_probability:.4f}")
print(f"Probability of a profit on Wednesday: {wed_profit_probability:.4f}")
```
