```python
def label_encode(s, order=None):
    cats = order or sorted(s.dropna().unique())
    mapping = {c: i for i, c in enumerate(cats)}
    return s.map(mapping), mapping


def one_hot_encode(s, prefix=None):
    prefix = prefix or s.name
    cats = sorted(s.dropna().unique())
    return pd.DataFrame({f"{prefix}_{c}": (s == c).astype(int) for c in cats})


edu, edu_map = label_encode(df["Education"], ["Basic", "2n Cycle", "Graduation", "Master", "PhD"])
print(edu_map)
one_hot_encode(df["Marital_Status"], "Marital").head()
```

```python
def encode_dataset(df):
    out = df.drop(columns=["ID"]).copy()
    out["Education"], _ = label_encode(
        out["Education"], ["Basic", "2n Cycle", "Graduation", "Master", "PhD"]
    )
    ohe = one_hot_encode(out["Marital_Status"], "Marital")
    out = pd.concat([out.drop(columns=["Marital_Status"]), ohe], axis=1)
    dates = pd.to_datetime(out["Dt_Customer"], dayfirst=True, errors="coerce")
    dates = dates.fillna(pd.to_datetime(out["Dt_Customer"], errors="coerce"))
    out["Dt_Customer"] = (dates - dates.min()).dt.days
    return out.fillna(out.median(numeric_only=True))


enc = encode_dataset(df)
X = enc.to_numpy(float)
print(f"Original: {df.shape}  Encoded: {enc.shape}")
enc.head()
```

```python
def minkowski(a, b, p=2):
    a, b = np.asarray(a, float), np.asarray(b, float)
    return float(np.sum(np.abs(a - b) ** p) ** (1 / p))


print("p=1:", minkowski(X[0], X[1], 1))
print("p=2:", minkowski(X[0], X[1], 2))
```

```python
ps = range(1, 11)
dists = [minkowski(X[0], X[1], p) for p in ps]
print(list(zip(ps, dists)))

plt.plot(ps, dists, marker="o")
plt.xlabel("p"); plt.ylabel("distance"); plt.title("Minkowski vs p")
plt.grid(True); plt.show()
```

```python
def dot(a, b):
    return float(sum(x * y for x, y in zip(a, b)))


def length(a):
    return float(sum(x * x for x in a) ** 0.5)


a, b = X[0], X[1]
print("dot :", dot(a, b), np.dot(a, b))
print("||a||:", length(a), np.linalg.norm(a))
print("||b||:", length(b), np.linalg.norm(b))
```

```python
def mean(x):
    x = np.asarray(x, float)
    return float(x.sum() / len(x))


def variance(x):
    x = np.asarray(x, float)
    m = mean(x)
    return float(((x - m) ** 2).sum() / len(x))


def std(x):
    return variance(x) ** 0.5


def feature_stats(X, names):
    return pd.DataFrame({
        "mean": [mean(X[:, j]) for j in range(X.shape[1])],
        "var":  [variance(X[:, j]) for j in range(X.shape[1])],
        "std":  [std(X[:, j]) for j in range(X.shape[1])],
    }, index=names)


stats = feature_stats(X, enc.columns)
stats.head(10)
```

```python
cmp = pd.DataFrame({
    "custom_mean": stats["mean"],
    "numpy_mean": np.mean(X, axis=0),
    "custom_std": stats["std"],
    "numpy_std": np.std(X, axis=0),
})
print("max mean diff:", (cmp["custom_mean"] - cmp["numpy_mean"]).abs().max())
print("max std  diff:", (cmp["custom_std"] - cmp["numpy_std"]).abs().max())
cmp.head(10)
```

## K-Means Clustering

This section implements K-Means manually using the algorithm steps from the lab:

1. Choose `k` initial centroids.
2. Assign every point to the nearest centroid.
3. Recalculate each centroid using the mean of its assigned points.
4. Repeat until the centroids stop changing or the maximum iterations are reached.

The features are scaled first because K-Means is distance-based.

```python
# Response is the target column, so it is removed from unsupervised clustering.
cluster_features = enc.drop(columns=["Response"], errors="ignore")
raw_cluster_matrix = cluster_features.to_numpy(float)

# Manual standardization: z = (x - mean) / standard deviation
feature_means = raw_cluster_matrix.mean(axis=0)
feature_stds = raw_cluster_matrix.std(axis=0)
feature_stds[feature_stds == 0] = 1
scaled_features = (raw_cluster_matrix - feature_means) / feature_stds


def euclidean_distance_matrix(points, centroids):
    differences = points[:, np.newaxis, :] - centroids[np.newaxis, :, :]
    return np.sqrt(np.sum(differences ** 2, axis=2))


def initialize_centroids(points, k, random_state=42):
    rng = np.random.default_rng(random_state)
    selected_indices = rng.choice(points.shape[0], size=k, replace=False)
    return points[selected_indices].copy()


def assign_nearest_cluster(points, centroids):
    distances = euclidean_distance_matrix(points, centroids)
    return np.argmin(distances, axis=1)


def update_centroids(points, labels, k, old_centroids):
    new_centroids = old_centroids.copy()
    for cluster_id in range(k):
        cluster_points = points[labels == cluster_id]
        if len(cluster_points) > 0:
            new_centroids[cluster_id] = cluster_points.mean(axis=0)
    return new_centroids


def calculate_inertia(points, labels, centroids):
    total_distance = 0
    for cluster_id, centroid in enumerate(centroids):
        cluster_points = points[labels == cluster_id]
        total_distance += np.sum((cluster_points - centroid) ** 2)
    return float(total_distance)


def manual_kmeans(points, k, max_iterations=100, tolerance=1e-4, random_state=42):
    centroids = initialize_centroids(points, k, random_state)

    for iteration in range(max_iterations):
        labels = assign_nearest_cluster(points, centroids)
        new_centroids = update_centroids(points, labels, k, centroids)
        centroid_shift = np.linalg.norm(new_centroids - centroids)
        centroids = new_centroids

        if centroid_shift < tolerance:
            break

    final_labels = assign_nearest_cluster(points, centroids)
    inertia = calculate_inertia(points, final_labels, centroids)
    return final_labels, centroids, inertia, iteration + 1


k_values = range(1, 9)
inertia_values = []

for k in k_values:
    _, _, inertia, _ = manual_kmeans(scaled_features, k, random_state=42)
    inertia_values.append(inertia)

plt.plot(list(k_values), inertia_values, marker="o")
plt.xlabel("Number of clusters (k)")
plt.ylabel("Within-cluster sum of squares")
plt.title("Elbow Method Using Manual K-Means")
plt.grid(True)
plt.show()
```

```python
# Choose k from the elbow plot. Change this value if your lab question specifies another k.
selected_k = 3

final_cluster_ids, final_centroids, final_inertia, total_iterations = manual_kmeans(
    scaled_features,
    selected_k,
    random_state=42,
)

print("Selected k:", selected_k)
print("Iterations used:", total_iterations)
print("Final within-cluster sum of squares:", round(final_inertia, 2))

clustered_df = df.copy()
clustered_df["Cluster"] = final_cluster_ids
enc_with_clusters = enc.copy()
enc_with_clusters["Cluster"] = final_cluster_ids

# Plot clusters using two easy-to-understand variables.
total_spending = clustered_df[
    [
        "MntWines",
        "MntFruits",
        "MntMeatProducts",
        "MntFishProducts",
        "MntSweetProducts",
        "MntGoldProds",
    ]
].sum(axis=1)

plt.figure(figsize=(7, 5))
plt.scatter(clustered_df["Income"], total_spending, c=final_cluster_ids, cmap="viridis", s=25)
plt.xlabel("Income")
plt.ylabel("Total Spending")
plt.title("Manual K-Means Customer Clusters")
plt.colorbar(label="Cluster")
plt.show()

profile_columns = [
    "Income",
    "Recency",
    "MntWines",
    "MntFruits",
    "MntMeatProducts",
    "MntFishProducts",
    "MntSweetProducts",
    "MntGoldProds",
    "NumWebPurchases",
    "NumCatalogPurchases",
    "NumStorePurchases",
    "Response",
]

cluster_profile = clustered_df.groupby("Cluster")[profile_columns].mean().round(2)
cluster_profile["Customer_Count"] = clustered_df.groupby("Cluster").size()
cluster_profile
```
